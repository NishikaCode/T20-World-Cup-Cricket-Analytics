```python
#17: Har team ke total runs calculate karo
team_scores = (
    df_clean.groupby(
        ["matchID", "match", "currentInning"],
        as_index=False
    )
    .agg(
        Total_Runs=("runs", "sum"),
        Wickets=("isWicket", "sum")
    )
)

team_scores = team_scores.sort_values(
    "Total_Runs", ascending=False
)

team_scores.head(10)
```

```python
#Highest-scoring innings ka graph banao
top_innings = team_scores.nlargest(10, "Total_Runs")

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_innings,
    x="currentInning",
    y="Total_Runs"
)

plt.title("Top 10 Highest-Scoring Innings")
plt.xlabel("Batting Team")
plt.ylabel("Total Runs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
#Top run scorers calculate karo
top_run_scorers = (
    df_clean.groupby("batsmanName")
    .agg(
        Total_Runs=("batsmanRuns", "sum"),
        Balls_Faced=("batsmanBall", "sum")
    )
    .sort_values("Total_Runs", ascending=False)
)

top_run_scorers.head(10)
```

```python
#Top 10 run scorers ka graph
top10 = top_run_scorers.head(10).reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(data=top10, x="batsmanName", y="Total_Runs")

plt.title("Top 10 Run Scorers - T20 World Cup 2024")
plt.xlabel("Player")
plt.ylabel("Total Runs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
#Top 10 wicket-takers ka analysis
bowler_wickets = (
    df_clean[df_clean["isBowlerWicket"] == True]
    .groupby("wktBowlerName")
    .size()
    .reset_index(name="Wickets")
    .sort_values("Wickets", ascending=False)
)

bowler_wickets.columns = ["Bowler", "Wickets"]

bowler_wickets.head(10)
```

```python
#Wicket-takers ka graph
top_bowlers = bowler_wickets.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_bowlers,
    x="Bowler",
    y="Wickets"
)

plt.title("Top 10 Wicket-Takers - T20 World Cup 2024")
plt.xlabel("Bowler")
plt.ylabel("Wickets")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
#eam-wise total runs ka graph
team_totals = (
    df_clean.groupby("currentInning")["runs"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

team_totals.columns = ["Team", "Total_Runs"]

plt.figure(figsize=(12, 6))

sns.barplot(
    data=team_totals,
    x="Team",
    y="Total_Runs"
)

plt.title("Total Runs by Team - T20 World Cup 2024")
plt.xlabel("Team")
plt.ylabel("Total Runs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
#Batting strike rate calculate karo
player_stats = (
    df_clean.groupby("batsmanName")
    .agg(
        Runs=("batsmanRuns", "sum"),
        Balls_Faced=("batsmanBall", "sum")
    )
    .reset_index()
)

player_stats["Strike_Rate"] = (
    player_stats["Runs"] / player_stats["Balls_Faced"] * 100
).where(player_stats["Balls_Faced"] > 0)

player_stats = player_stats.sort_values(
    "Runs", ascending=False
)

player_stats.head(10)
```
